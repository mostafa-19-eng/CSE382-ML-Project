```python
import pandas as pd
from pathlib import Path

data_path = Path("../data/H2.csv")
df = pd.read_csv(data_path)

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())
```

```python
# 1. Check column names and data types
print("COLUMN NAMES AND DATA TYPES")
df.info()

# 2. Check missing values
print("\nMISSING VALUES")
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))

# 3. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# 4. Check basic statistics
print("\nNUMERICAL SUMMARY")
display(df.describe().T)
```

```python

print("MISSING VALUES")
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

print("\nUNIQUE VALUES IN IMPORTANT COLUMNS")
for col in ["IsCanceled", "Children", "Meal", "MarketSegment",
            "DepositType", "CustomerType", "ReservationStatus"]:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False).head(10))
```

```python

print("CATEGORICAL COLUMN SUMMARY")

for col in ["Meal", "MarketSegment", "DepositType",
            "CustomerType", "ReservedRoomType", "AssignedRoomType"]:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).head(10))

print("\nCHECK ADR EXTREME VALUES")
display(
    df.nlargest(10, "ADR")[
        ["ADR", "Adults", "Children", "Babies",
         "MarketSegment", "DepositType", "IsCanceled"]
    ]
)

print("\nCHECK ZERO OR NEGATIVE ADR")
print("ADR <= 0:", (df["ADR"] <= 0).sum())
```
